# 01. Financial Document Parsing & Table Extraction EDA

This notebook demonstrates how `PyMuPDF4LLM` extracts corporate 10-K filings into clean Markdown tables while preserving column alignment, numerical integrity, and fiscal metadata.

```python
from pathlib import Path
import sys
sys.path.append('..')

from src.ingestion.parser import FinancialDocumentParser
from src.ingestion.chunking import FinancialChunker

# 1. Parse sample PDF filing
sample_pdf = Path('../data/raw/apple_10k_sample.pdf')
parser = FinancialDocumentParser()
pages = parser.parse_pdf(sample_pdf)

print(f'Total pages parsed: {len(pages)}')
print('Page 2 Preview (Financial Statement Table):')
print(pages[1]['content'][:400])
```

```python
# 2. Chunking with Table Header Preservation
chunker = FinancialChunker(chunk_size=1200, chunk_overlap=150)
output_jsonl = Path('../data/processed/apple_10k_sample.jsonl')
chunks = chunker.process_and_save(pages, output_jsonl)

print(f'Generated {len(chunks)} searchable chunks.')
for c in chunks:
    print(f"Chunk ID: {c['chunk_id']} | Page: {c['page_number']} | Has Table: {c['has_table']}")
```
